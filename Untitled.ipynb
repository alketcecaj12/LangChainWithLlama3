```python
# simple_rag.py

import numpy as np
from sentence_transformers import SentenceTransformer
import requests
import json

# ─── CONFIG ────────────────────────────────────────────────────────────────
OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL = "llama3.2"
EMBED_MODEL = "all-MiniLM-L6-v2"  # small, fast, local embedding model
TOP_K = 2  # how many chunks to retrieve

# ─── DOCUMENT STORE ────────────────────────────────────────────────────────
# Replace these with your own documents / chunks
DOCUMENTS = [
    "Nordea is a Nordic financial services group headquartered in Helsinki, Finland.",
    "Credit risk refers to the risk that a borrower will default on their debt obligations.",
    "IFRS 9 is an accounting standard that requires banks to recognize expected credit losses.",
    "PD stands for Probability of Default. It measures the likelihood a borrower will default.",
    "LGD stands for Loss Given Default. It measures the loss if a borrower does default.",
    "EAD stands for Exposure at Default. It is the total value exposed to loss at the time of default.",
    "Apache Airflow is an open-source workflow orchestration tool used for data pipelines.",
    "dbt (data build tool) is used for transforming data in a data warehouse like Snowflake.",
]

# ─── STEP 1: EMBED DOCUMENTS ───────────────────────────────────────────────
print("Loading embedding model...")
embedder = SentenceTransformer(EMBED_MODEL)
doc_embeddings = embedder.encode(DOCUMENTS, convert_to_numpy=True)
print(f"Embedded {len(DOCUMENTS)} documents.\n")

# ─── STEP 2: RETRIEVER ─────────────────────────────────────────────────────
def retrieve(query: str, top_k: int = TOP_K) -> list[str]:
    query_embedding = embedder.encode([query], convert_to_numpy=True)
    
    # Cosine similarity
    norms = np.linalg.norm(doc_embeddings, axis=1) * np.linalg.norm(query_embedding)
    similarities = np.dot(doc_embeddings, query_embedding.T).flatten() / norms
    
    top_indices = np.argsort(similarities)[::-1][:top_k]
    return [DOCUMENTS[i] for i in top_indices]

# ─── STEP 3: GENERATE WITH CONTEXT ─────────────────────────────────────────
def ask(question: str) -> str:
    context_chunks = retrieve(question)
    context = "\n".join(f"- {c}" for c in context_chunks)
    
    # Keep the prompt tight for smaller models
    prompt = f"""Use only the context below to answer the question. Be concise.

Context:
{context}

Question: {question}
Answer:"""

    response = requests.post(OLLAMA_URL, json={
        "model": MODEL,
        "prompt": prompt,
        "stream": False
    })
    return response.json()["response"].strip()

```

```python
# ─── MAIN LOOP ──────────────────────────────────────────────────────────────
if __name__ == "__main__":
    print("Simple RAG with LLaMA 3.2 + Ollama")
    print("Type 'quit' to exit.\n")
    while True:
        question = input("You: ").strip()
        if question.lower() in ("quit", "exit"):
            break
        answer = ask(question)
        print(f"RAG: {answer}\n")
```
